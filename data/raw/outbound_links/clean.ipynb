```python
import pandas as pd
from urllib.parse import urlparse

def get_domain(url):
    try:
        parsed = urlparse(url)
        netloc = parsed.netloc
        if not netloc:
            return ""
        parts = netloc.split('.')
        if len(parts) >= 2:
            return '.'.join(parts[-2:])
        else:
            return netloc
    except:
        return ""

df = pd.read_csv('outbound_links_top_domain.csv')
print(f"Jumlah baris sebelum filtering: {len(df)}")
df2 = pd.read_csv('../raw/data_top_domain.csv')

df['clean_domain'] = df['url'].apply(get_domain)
df = df.drop_duplicates(subset=['clean_domain'], keep='first')

existing_domains = df2['url'].apply(get_domain).unique()

df = df[~df['clean_domain'].isin(existing_domains)]
print(f"Jumlah baris setelah filtering: {len(df)}")
df = df.drop(columns=['clean_domain'])

df.to_csv('outbound_links_top_domain_clean.csv', index=False)
```
