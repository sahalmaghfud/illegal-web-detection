```python
import pandas as pd

df1 = pd.read_csv('data_top_domain.csv')
df2 = pd.read_csv('data_top_domain_outbound_links.csv')
df3 = pd.read_csv('data_serp.csv')
df4 = pd.read_csv('data_serp_outbound_links.csv')

df = pd.concat([df1, df2, df3, df4], ignore_index=True)
```

```python
import os
from urllib.parse import urlparse

def get_domain(url):
    try:
        parsed = urlparse(str(url))
        domain = parsed.netloc.replace('www.', '')
        return domain.lower()
    except:
        return url

df['clean_domain'] = df['url'].apply(get_domain)
df = df.drop_duplicates(subset=['clean_domain'], keep='first')

# cek file fisik
indices_to_drop = []

for index, row in df.iterrows():
    img_file = "../" + row['image_path']
    html_file = "../" + row['html_path']

    exists_img = os.path.isfile(img_file)
    exists_html = os.path.isfile(html_file)

    if not exists_img or not exists_html:
        indices_to_drop.append(index)
    
df_final = df.drop(indices_to_drop)


df_final = df_final.drop(columns=['clean_domain'])

output_file = "data_unlabeled.csv"
df_final.to_csv(output_file, index=False)

print(f"Data sebelumnya: {len(df)}")
print(f"Data setelah filtering: {len(df_final)}")
```
