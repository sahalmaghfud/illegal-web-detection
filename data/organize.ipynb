```python
import pandas as pd
import os
import shutil

csv_name = 'labeled_data.csv'

df = pd.read_csv(csv_name)
df = df.drop(df[df['label'] == 4].index)
print(df['label'].value_counts())


img_dir = 'dataset_image'
html_dir = 'dataset_html'

def copy_and_update(row, path_col, base_new_dir):
    old_path = row[path_col]
    
    if pd.isna(row['label']):
        return old_path
        
    label = str(int(row['label']))
    
    if pd.isna(old_path) or not isinstance(old_path, str):
        return old_path
    
    target_dir = os.path.join(base_new_dir, label)
    os.makedirs(target_dir, exist_ok=True)
    
    file_name = os.path.basename(old_path)
    new_path = os.path.join(target_dir, file_name)
    
    try:
        if os.path.exists(old_path):
            shutil.copy2(old_path, new_path)
            return new_path 
        else:
            return old_path
    except Exception:
        return old_path

df['image_path'] = df.apply(lambda r: copy_and_update(r, 'image_path', img_dir), axis=1)
df['html_path'] = df.apply(lambda r: copy_and_update(r, 'html_path', html_dir), axis=1)

output_csv = 'labeled_data_final.csv'
df.to_csv(output_csv, index=False)

print(f"\nSelesai! Dataframe baru disimpan di: {output_csv}")
```
